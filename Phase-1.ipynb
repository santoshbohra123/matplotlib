```python
import numpy as np

x = ["Python", "C", "C++", "Java"]
y = [85,70,60,80]
z = [20,30,40,50]
width = 0.2

# p = np.arange(5)
p = np.arange(len(x))
# print(p)
p1 = [j+width for j in p]  # For every value j in p, add width to it.

plt.xlabel("Programming Languages",fontsize = 15)
plt.ylabel("Numbers",fontsize = 15)
plt.title("Data",fontsize = 15)
c= ["y","b","m","g"]    # provide color code and name to the color paramete in below
plt.bar(x, z, width, color="r", align="center", edgecolor="black",label = "popularity2")
plt.bar(p1, y, width, color="g", align="center", edgecolor="black",label = "popularity1")
plt.legend()  # use labels using legend function. and we can provide labels inside legend func as params.
# plt.bar(x,y,width=0.4,color = c, align="center",edgecolor = "r",linewidth = 5,linestyle = ":",alpha = 0.5) 

plt.xticks(p+width/2,x,rotation = 50)
# Syntax of xticks()=> plt.xticks(ticks, labels)
# ticks → Where the labels should appear on the x-axis.
# labels → What text should be displayed there.


# parameters like width and color
# align by default use center we can provide it edge.
# value of alpha is between 0 and 1.
plt.show()
```

# Horizontal bar graph

```python
import numpy as np

x = ["Python", "C", "C++", "Java"]
y = [85,70,60,80]
z = [20,30,40,50]
width = 0.2

# p = np.arange(5)
p = np.arange(len(x))
# print(p)
p1 = [j+width for j in p]  # For every value j in p, add width to it.

plt.xlabel("Programming Languages",fontsize = 15)
plt.ylabel("Numbers",fontsize = 15)
plt.title("Data",fontsize = 15)
# use numeric y positions (p and p1) for barh and set height explicitly
plt.barh(p, z, height=width, color="r", align="center", edgecolor="black", label="popularity2")
plt.barh(p1, y, height=width, color="g", align="center", edgecolor="black", label="popularity1")
plt.legend()  # use labels using legend function. and we can provide labels inside legend func as params.
# plt.bar(x,y,width=0.4,color = c, align="center",edgecolor = "r",linewidth = 5,linestyle = ":",alpha = 0.5) 

# set y-ticks at the middle of grouped bars and use category labels
plt.yticks(p + width/2, x, rotation = 50)
# Syntax of yticks()=> plt.yticks(ticks, labels)
# ticks → Where the labels should appear on the y-axis.
# labels → What text should be displayed there.


# parameters like width and color
# align by default use center we can provide it edge.
# value of alpha is between 0 and 1.
plt.show()
```
