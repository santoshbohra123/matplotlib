```python
# Surface plot

x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)

xx,yy = np.meshgrid(x,y)
zz = xx**2 + yy**2

fig = plt.figure(figsize = (12,8))
ax = plt.subplot(projection = "3d")
p = ax.plot_surface(xx,yy,zz,cmap = "viridis")

ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("z")

fig.colorbar(p)
```

```python
z = np.sin(xx) + np.cos(yy)
fig = plt.figure(figsize = (12,8))
ax = plt.subplot(projection = "3d")
p = ax.plot_surface(xx,yy,z,cmap = "viridis")

ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("z")

fig.colorbar(p)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Data
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)

xx, yy = np.meshgrid(x, y)

zz = xx**2 + yy**2
z = np.sin(xx) + np.cos(yy)

# Figure
fig = plt.figure(figsize=(15,10))

# ---------------- Surface Plot 1 ----------------
ax1 = fig.add_subplot(2,2,1, projection='3d')
ax1.plot_surface(xx, yy, zz, cmap='viridis')
ax1.set_title("z = x² + y²")

# ---------------- Contour Plot 1 ----------------
ax2 = fig.add_subplot(2,2,2)
ax2.contourf(xx, yy, zz, cmap='viridis')
ax2.set_title("Contour : x² + y²")

# ---------------- Surface Plot 2 ----------------
ax3 = fig.add_subplot(2,2,3, projection='3d')
ax3.plot_surface(xx, yy, z, cmap='plasma')
ax3.set_title("z = sin(x) + cos(y)")

# ---------------- Contour Plot 2 ----------------
ax4 = fig.add_subplot(2,2,4)
ax4.contourf(xx, yy, z, cmap='plasma')
ax4.set_title("Contour : sin(x) + cos(y)")

plt.tight_layout()
plt.show()
```

```python
temp_df = delivery[(delivery["ballnumber"].isin([1,2,3,4,5,6])) & (delivery["batsman_run"] == 6 )]  # Boolean Indexing kari hai.

# Generally isin() ko filter ke liye use karte hain. take1  argument list, set or tuple. return bool True and false value based on argument.
#  it checks "Har row ka value list me hai ya nahi?"

grid = temp_df.pivot_table(index= "overs", columns= 'ballnumber',values = "batsman_run", aggfunc='count')
```
