```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

# Colour incoding in 2D scatter plot.

```python

iris = pd.read_csv("/content/iris.csv")
iris.sample(5)
```

```python
iris['Species'] = iris['Species'].replace({"Iris-versicolor":0,'Iris-setosa':1,'Iris-virginica':2})
iris.sample(5)
```

```python
plt.scatter(iris["SepalLengthCm"],iris["PetalLengthCm"],c=iris['Species'],cmap='jet',alpha = 0.8) # c is here hidden parameter->it provide color to diff classes.
# cmap is color map provide cmaps(color maps).

plt.title("Iris Dataset")
plt.xlabel("SepalLengthCm")
plt.ylabel("PetalLengthCm")
plt.colorbar()  # it tells about which color represent which value  red = 2, green = 0 , blue = 1. now we can easily watch their replaceble names.
plt.show()
```
