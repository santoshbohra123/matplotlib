```python
iris['Species'] = iris['Species'].replace({"Iris-versicolor":0,'Iris-setosa':1,'Iris-virginica':2})
iris.sample(5)
```

```python
plt.scatter(iris["SepalLengthCm"],iris["PetalLengthCm"],c=iris['Species'],cmap='jet',alpha = 0.8) # c is here hidden parameter->it provide color to diff classes.
# cmap is color map provide cmaps(color maps).

plt.title("Iris Dataset")
plt.xlabel("SepalLengthCm")
plt.ylabel("PetalLengthCm")
plt.colorbar()  # it tells about which color represent which value  red = 2, green = 0 , blue = 1. now we can easily watch their replaceble names.
plt.show()
```

```python
batters = pd.read_csv("/content/drive/MyDrive/datasets/batter (1).csv")
```

```python
fig,ax = plt.subplots(figsize= (15,6))
ax.scatter(batters['avg'],batters['strike_rate'],color = "red")
ax.set_xlabel("avg")
ax.set_ylabel("strike_rate")
ax.set_title("batter")

fig.show()
```

```python
# 2 scatter plots in one using sub plots
# 1- strike_rate vs runs
# 2 - avg vs strike_rate
```

```python
fig,ax = plt.subplots(nrows=2,ncols=1,sharex = True,figsize=(15,10))
# print(fig)
# print(ax)

ax[0].scatter(batters["avg"],batters["runs"],color = 'red')

ax[0].set_title("avg vs runs")
# ax[0].set_xlabel("avg")
ax[0].set_ylabel("runs")

ax[1].scatter(batters["avg"],batters["strike_rate"])

ax[1].set_title("avg vs strike rate")
ax[1].set_xlabel("avg")
ax[1].set_ylabel("strike rate")

fig.show()
```

```python
fig,ax = plt.subplots(nrows=2,ncols=2,figsize=(15,10))
# print(fig)
# print(ax)

ax[0][0].scatter(batters["avg"],batters["runs"],color = 'red')

ax[0][0].set_title("avg vs runs")
# ax[0].set_xlabel("avg")
ax[0][0].set_ylabel("runs")

ax[0][1].scatter(batters["avg"],batters["strike_rate"])

ax[0][1].set_title("avg vs strike rate")
ax[0][1].set_xlabel("avg")
ax[0][1].set_ylabel("strike rate")

ax[1][0].hist(batters["runs"])

ax[1][0].set_title("Batter's runs")
ax[1][0].set_xlabel("runs")


ax[1][1].hist(batters["avg"])

ax[1][1].set_title("Batter's avg ")
ax[1][1].set_xlabel("avg")


fig.show()
```

```python
# Another ways to create graph

# fig = plt.figure()
fig = plt.figure(figsize=(12,8))

ax1 = fig.add_subplot(2,2,1)
ax1.scatter(batters["avg"],batters["runs"],color = 'red')

ax1.set_title("avg vs runs")
ax1.set_xlabel("avg")
ax1.set_ylabel("runs")

ax2 = fig.add_subplot(2,2,2)
ax2.scatter(batters["avg"],batters["strike_rate"],color = 'red')
ax2.set_title("avg vs strike_rate")
ax2.set_xlabel("avg")
ax2.set_ylabel("strike rate")

ax3 = fig.add_subplot(2,2,3)
ax3.hist(batters["runs"])
ax3.set_title("Batter's runs")
ax3.set_xlabel("runs")

# for spacing

plt.tight_layout()
plt.show()

# fig.subplots_adjust(
#     hspace=0.5,   # Vertical spacing
#     wspace=0.3    # Horizontal spacing
# )

# plt.show()
```
