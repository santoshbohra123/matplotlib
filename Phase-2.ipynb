# Bar graph

bivariare analysis and univariate analysis
Numerical (generally aggregate) vs categorical
use case - aggregate analysis of group.


```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
children = [10,20,30,25,18]
color = ["Red", "green","yellow","pink","brown"]


plt.bar(color,children,color="black")
```

```python
children = [10,20,30,25,18]
color = ["Red", "green","yellow","pink","brown"]


plt.barh(color,children,color="black")
```

```python
df = pd.read_csv("batsman_season_record.csv")

width = 0.27
p = np.arange(len(df["batsman"]))
p1 = [j+width for j in p]
p2 = [j-width for j in p]

plt.bar(p1,df["2015"],width)
plt.bar(p,df["2016"],width)
plt.bar(p2,df["2017"],width)

plt.xticks(p,df["batsman"],rotation = 50)

plt.show()
```

```python
# Stacked bar chart
plt.bar(df["batsman"],df["2015"], label = "2015")
plt.bar(df["batsman"],df["2016"],bottom = df["2015"], label = "2016")
plt.bar(df["batsman"],df["2017"],bottom = df["2015"]+ df["2016"], label = "2017")
plt.legend()
plt.show()
```
