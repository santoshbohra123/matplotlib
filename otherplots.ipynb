```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5,6]
y = [2,6,5,7,3,8]

plt.stem(x,y,linefmt=':',markerfmt='ro',bottom=6,basefmt='g',label='Python',)
plt.legend()
# basefmt - change base line color
# linefmt - change line style
# markerfmt - change marker color
#orientation='horizontal' change vertical to horizental.

plt.show()
```

```python
# Box Plot and whisker plot
x = [10,20,30,40,50,60,70,120]
# median = 40 make/draw median line
# q1 = (25)1st quartile , q2 = median and q3 = (55) 2nd quartile
# 10 = minimum value and 70 = max value
# only q1 to q3 mart is box plot and 
# q1 to min-  line and q3 to max line called wiskey plot
# wiskey plot show min and maximum of box plot.
# o shape in box plot means value is far from it.
# max and minimum ki line ko CAP bolte hn
```

```python
plt.boxplot(x,vert=True,widths=0.2,label=["python"],showmeans=True,whis=2,sym='g+',boxprops=dict(color = 'r'),
            capprops=dict(color = 'g'),whiskerprops=dict(color = 'r'))
# notch= True shape change 
# vert=False horizental boxplot
# label always parameters as a list pass.
# patch_artist=True  it fills color inside
# showmeans=True -> mean visible
# whis=2   wisk join with outlier
# outlier hide   sym=' '
# boxprops=dict(color = 'r'),capprops=dict(color = 'g') ->   both takes dict()

plt.legend()
plt.show()
```
